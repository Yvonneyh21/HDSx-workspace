```python
import pandas as pd
```

```python
nhanes = pd.read_csv(
     "../../../examples/nhanes-equity/data/nhanes_equity_v6.csv"
)

print(nhanes.shape)
print(nhanes.columns.tolist())
```

```python

nhanes_clean = nhanes.dropna(
    subset=["BMI", "IncomeGroup", "Gender"]
)

print(len(nhanes_clean))
```

```python
bmi_summary_py = (
    nhanes_clean
    .groupby(["IncomeGroup", "Gender"])["BMI"]
    .agg(["count", "mean", "std"])
    .reset_index()
)

bmi_summary_py
```

```python
from helpers import summarize_bmi

bmi_helper_result = summarize_bmi(nhanes)

bmi_helper_result
```

```python
bmi_summary_py.equals(bmi_helper_result)
```

```python
bmi_summary_rounded_py = bmi_summary_py.copy()

bmi_summary_rounded_py["mean"] = bmi_summary_rounded_py["mean"].round(2)
bmi_summary_rounded_py["std"] = bmi_summary_rounded_py["std"].round(2)

bmi_summary_rounded_py
```
